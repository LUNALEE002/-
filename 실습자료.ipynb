```python
import torch

assert torch.cuda.is_available(), "GPU가 꺼져있는데요?"

dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16

print("GPU 준비 완료")


```

```python
from qwen_tts import Qwen3TTSModel

model=Qwen3TTSModel.from_pretrained("Qwen/Qwen3-TTS-12Hz-0.6B-Base",device_map="cuda:0", dtype=dtype,)

print("모델 로드 완료!")

```

```python
from google.colab import files

print("당신의 목소리를 3초 이상 녹음해서 올리세요!")

uploaded= files.upload()

내목소리 = list(uploaded.keys())[0]

print("내 목소리 업로드 완료하였습니다!")
```

```python
import soundfile as sf
import importlib
importlib.reload(sf)
from IPython.display import Audio, display

녹음내용="안녕하세요 저는 루나라고 합니다. 저는 인공지능을 배우고 있어요. 저는 인공지능을 너무너무 사랑합니다."
생성내용="""국가 최고 기밀, 지하 50층에 잠든 비밀 벙커의 존재. 과연 그 안에는 무엇이 숨겨져 있을까요?
겉으로는 평범한 건축물 같지만, 그 구조는 극한의 압력을 견디기 위해 설계된 치명적인 비밀입니다.
50층이라는 압도적인 깊이, 땅의 모든 힘을 담아 짓눌러 오는 물리적 위협에 맞서야 했습니다.
정밀하게 계산된 압력 수치, 설계 상의 한계를 초과하는 순간 찾아온 치명적인 재앙이었습니다.
문제는 단순히 뚫거나 치우는 것으로 해결될 수 없는, 구조적인 딜레마 그 자체였습니다.
외부의 압력과 내부 환경이 끊임없이 상충하며 벙커를 서서히 무너뜨리는 역설적인 싸움이었죠.
이에 전문가들은 역발상을 통해 해결책을 찾았습니다. 핵심은 '힘의 분산'이었습니다.
특수한 재료와 분산 구조를 결합해, 압력을 흡수하는 동시에 안정성을 확보하는 기적의 해법이 나왔습니다.
이 이야기는 단순한 미스터리가 아닙니다. 인간의 탐구심과 기술의 한계를 보여주는 씁쓸한 통찰입니다.
극한의 환경 속에서 우리는 무엇을 배웠을까요? 다음 다큐멘터리에서 만나요."""

wavs, sr = model.generate_voice_clone( text = 생성내용,
                                      language="Korean",
                                       ref_audio=내목소리,
                                       ref_text=녹음내용,
                                       )
sf.write("내목소리.wav", wavs[0], sr)
print("생성완료!!")

display(Audio("내목소리.wav"))
```
